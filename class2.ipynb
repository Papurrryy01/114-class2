# Stacks

A stack is an ordered collection of elements where itmes are added and removed from one end, comonly referred to as the "top". The ordering principleis referred to as LIFO(Last in, Last out)

# QUEUES



```python
stack = []

stack.append("A")
stack.append("B")
stack.append("C")

top = stack.pop()

print(top, "has been removed from stack")  # Outputs: C
print("updated stack",stack)
print("Removed top element from stack:", top)

next_val = stack[-1]
print("Next top element is:", next_val)  # Outputs: B

# Push D
stack.append("D")
print("updated stack",stack) # Outputs: ['A', 'B', 'D']

# Pop D and B
print(stack.pop(), "has been removed from stack")  # Outputs: D
print(stack.pop(), "has been removed from stack")  # Outputs: B
print("Final stack",stack)

# 2nd challenge

stack.append("1")
stack.append("2")
stack.append("3")
stack.append("4")
print("updated stack", stack)

print("Top element is:", stack[-1])  # Outputs: 4
```

```python
#reverse word

word = "hello"
stack = []

for letter in word:
    stack.append(letter)

reversed_word = ""
while len(stack) > 0:
    reversed_word += stack.pop()

print(reversed_word)
```
